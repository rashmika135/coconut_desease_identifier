### create the PyTorch Dataset

```python
### create the label mapping 
label_map = {'Healthy': 0,'CCI': 1,'WCLWD_Yellowing': 2,'WCLWD_Flaccidity': 3,'WCLWD_Drying': 4}
```

```python
from torch.utils.data import Dataset
class CoconutDataset(Dataset):

    def __init__(self, dataframe, transform):
        self.dataframe = dataframe
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):

        image_path = self.dataframe.iloc[index]['image_path']
        label = self.dataframe.iloc[index]['label']

        image = Image.open(image_path).convert('RGB')  
        image = self.transform(image)

        label = label_map[label]

        return image, label
```

```python
train_dataset = CoconutDataset(train_df, transform)
val_dataset = CoconutDataset(val_df, transform)
test_dataset = CoconutDataset(test_df, transform)
```

# Create the model

```python
import torch.nn as nn
class CoconutCNN(nn.Module):

    def __init__(self):
        super().__init__()

        self.conv1 = nn.Conv2d(3, 16, kernel_size=3)
        self.pool = nn.MaxPool2d(2, 2)

        self.conv2 = nn.Conv2d(16, 32, kernel_size=3)

        self.flatten = nn.Flatten()

        self.fc1 = nn.Linear(32 * 54 * 54, 128)
        self.fc2 = nn.Linear(128, 5)

        self.relu = nn.ReLU()

    def forward(self, x):

        x = self.pool(self.relu(self.conv1(x)))
        x = self.pool(self.relu(self.conv2(x)))

        x = self.flatten(x)

        x = self.relu(self.fc1(x))
        x = self.fc2(x)

        return x
```

```python
model = CoconutCNN()

print(model)
```

```python
epochs = 5

for epoch in range(epochs):

    model.train()

    running_loss = 0

    for images, labels in train_loader:

        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        running_loss += loss.item()

    average_loss = running_loss / len(train_loader)

    print('Epoch:', epoch + 1, 'Loss:', average_loss)
```

# ### Augmentation Results

Data augmentation was applied to improve the model's generalization, particularly for the Healthy class.

However, the results did not improve. The overall validation accuracy decreased from approximately 94% to 93%, while the Healthy class F1-score decreased from 0.65 to 0.53.

Therefore, basic data augmentation alone was not sufficient to solve the class imbalance problem. The main reason may be the very small number of Healthy training images compared with the other classes.

## Model Improvement 2 - Class Weights

The dataset is highly imbalanced, with significantly fewer Healthy images than the other classes.

Since data augmentation did not improve the Healthy class performance, class weights are used to give more importance to the minority class during training.

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.arange(5),
    y=train_df['label'].map(label_map)
)

print(class_weights)
```

```python
class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
)

print(class_weights)
```

```python
model = CoconutCNN()
criterion = nn.CrossEntropyLoss(
    weight=class_weights
)
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
epochs = 5

for epoch in range(epochs):

    model.train()
    running_loss = 0

    for images, labels in train_loader:

        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        running_loss += loss.item()

    average_loss = running_loss / len(train_loader)

    print('Epoch:', epoch + 1, 'Loss:', average_loss)
```

```python
model.eval()

all_predictions = []
all_labels = []

with torch.no_grad():

    for images, labels in val_loader:

        outputs = model(images)

        _, predicted = torch.max(outputs, 1)

        all_predictions.extend(predicted.tolist())
        all_labels.extend(labels.tolist())
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    all_labels,
    all_predictions
)

print('Validation Accuracy:', accuracy)

print(
    classification_report(
        all_labels,
        all_predictions,
        target_names=class_names
    )
)
```

### Class Weight Results

Class weights improved the Healthy class recall from 0.58 to 0.68 and slightly improved its F1-score from 0.65 to 0.67.

However, overall validation accuracy decreased from approximately 94% to 91%. The recall of WCLWD_Yellowing also decreased considerably.

Therefore, class weighting improved the minority Healthy class but reduced the overall balance of the model's performance.


## Transfer Learning with ResNet18

The baseline CNN achieved strong overall validation accuracy, but the Healthy class remained difficult because of the small number of training images.

To improve feature extraction, transfer learning is used with a pretrained ResNet18 model.

```python
from torchvision import models
model = models.resnet18(weights='DEFAULT')
model.fc = nn.Linear(model.fc.in_features, 5)
print(model.fc)

resnet_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

train_dataset = CoconutDataset(train_df, resnet_transform)
val_dataset = CoconutDataset(val_df, resnet_transform)
test_dataset = CoconutDataset(test_df, resnet_transform)

train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False
)

for param in model.parameters():
    param.requires_grad = False

for param in model.fc.parameters():
    param.requires_grad = True

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(
    model.fc.parameters(),
    lr=0.001
)
```

```python
epochs = 5

for epoch in range(epochs):

    model.train()
    running_loss = 0

    for images, labels in train_loader:

        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        running_loss += loss.item()

    average_loss = running_loss / len(train_loader)

    print('Epoch:', epoch + 1, 'Loss:', average_loss)
```

```python
model.eval()

all_predictions = []
all_labels = []

with torch.no_grad():

    for images, labels in val_loader:

        outputs = model(images)

        _, predicted = torch.max(outputs, 1)

        all_predictions.extend(predicted.tolist())
        all_labels.extend(labels.tolist())
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    all_labels,
    all_predictions
)

print('Validation Accuracy:', accuracy)
```

```python
print(
    classification_report(
        all_labels,
        all_predictions,
        target_names=class_names
    )
)
```

### Transfer Learning Results

The pretrained ResNet18 model significantly improved the classification performance.

The overall validation accuracy increased to approximately 97%. More importantly, the performance of the minority Healthy class improved substantially, with recall increasing from 0.58 in the baseline CNN to 0.89 and the F1-score increasing from 0.65 to 0.92.

The macro F1-score also increased from 0.89 to 0.96, showing better performance across all five classes.

Therefore, ResNet18 transfer learning provided the best validation performance among the tested approaches.`

```python
def predict_image(image_path):

    # Open the image
    image = Image.open(image_path).convert('RGB')

    # Apply the same preprocessing used by ResNet18
    image = resnet_transform(image)

    # Add batch dimension
    image = image.unsqueeze(0)

    # Make prediction
    model.eval()

    with torch.no_grad():

        output = model(image)

        probabilities = torch.softmax(output, dim=1)

        confidence, predicted = torch.max(probabilities, 1)

    # Get predicted class name
    predicted_class = class_names[predicted.item()]

    print('Prediction:', predicted_class)
    print('Confidence:', confidence.item() * 100, '%')
```
